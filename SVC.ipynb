```python
import kagglehub
from pyspark.sql import SparkSession
import pandas as pd
import os
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_auc_score, average_precision_score, precision_recall_curve, roc_curve, confusion_matrix, ConfusionMatrixDisplay
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline
```

```python
spark = SparkSession.builder.appName("FraudeSpark").getOrCreate()
```

```python
path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")
print("Path to dataset files:", path)
```

```python
file_path = os.path.join(path, "creditcard.csv")
df = spark.read.csv(file_path, header=True, inferSchema=True)
df.head()
```

```python
print(df['Class'].value_counts(normalize=True))
```

```python
X = df.drop(columns=['Class'])
y = df['Class']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=123, stratify=y)
```

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('smote', SMOTE(random_state=123)),
    ('model', SVC(random_state=123, probability=True))
])
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=123)
```

```python
param_grid = {
    'model__C': [0.1, 1],
    'model__kernel': ['linear', 'rbf'],
    'model__gamma': ['scale', 'auto']
}
```

```python
grid_search = GridSearchCV(
    pipeline,
    param_grid,
    scoring='recall',
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train, y_train)
best_model = grid_search.best_estimator_

print("\n🔥 Melhor parâmetros:")
print(grid_search.best_params_)
```

```python
importances = pd.Series(best_model.named_steps['model'].feature_importances_, index=best_model.feature_names_in_)
importances = importances.sort_values(ascending=False)

plt.figure(figsize=(10, 6))
importances.plot(kind='barh')
plt.title("Top Features")
plt.gca().invert_yaxis()
plt.show()
```

```python
y_pred = best_model.predict(X_test)
y_proba = best_model.predict_proba(X_test)[:, 1]
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues')
plt.title("Matriz de Confusão")
plt.show()
```

```python
roc_auc = roc_auc_score(y_test, y_proba)
pr_auc = average_precision_score(y_test, y_proba)

print(f"ROC-AUC: {roc_auc:.4f}")
print(f"PR-AUC: {pr_auc:.4f}")
```

```python
fpr, tpr, _ = roc_curve(y_test, y_proba)

plt.figure()
plt.plot(fpr, tpr)
plt.title("ROC Curve")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.show()
```

```python
precision, recall, _ = precision_recall_curve(y_test, y_proba)

plt.figure()
plt.plot(recall, precision)
plt.title("Precision-Recall Curve")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.show()
```

```python
prec, rec, thresholds = precision_recall_curve(y_test, y_proba)

f1_scores = 2 * (prec * rec) / (prec + rec)
threshold = thresholds[f1_scores.argmax()]
y_pred_custom = (y_proba >= threshold).astype(int)

print(classification_report(y_test, y_pred_custom))
```

```python
cm = confusion_matrix(y_test, y_pred_custom)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues')
plt.title("Matriz de Confusão")
plt.show()
```
